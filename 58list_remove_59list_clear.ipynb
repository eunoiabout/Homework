```python
#58리스트에서 요소 삭제하기: list.remove()
```

```python
numbers = [1,1,3,1,1,5,6,2,0,7]
if 1 in numbers:
    numbers.remove(1)
print(numbers)
```

```python
numbers = [1,1,3,1,1,5,6,2,0,7]
while 1 in range(numbers.count(1)):
    numbers.remove(1)
print(numbers)
```

```python
fruits = ['사과','배','오렌지','수박','참외']
fruits.remove('딸기')
```

```python
#59리스트의 모든 요소 삭제하기: list.clear()
```

```python
fruits = ['사과','배','오렌지','수박','참외']
fruits.clear()
print(fruits)
```

```python
fruits = ['사과','배','오렌지','수박','참외']
del fruits[:]
print(fruits)
```
