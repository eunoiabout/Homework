```python
#86 숫자 왼쪽에서 0을 채워서 문자열로 만들기:format()
```

```python
names = ['유비','관우','장비','조운','제갈량']
for idx, name in enumerate(names, start=1):
    order = format(idx,'03')
    print(f'{order}-{name}')
```
