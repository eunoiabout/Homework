```python
#90문자열에서 주어진 접두어/접미어 제거하기: str.removeprefix(), str.removesufffix()
```

```python
strData = '[info]Seoul is the captical of Korea.'
cityinfo = strData.removeprefix('[info]')
print(cityinfo)
```

```python
strData = '이것으로 문장을 마무리합니다.<EOD>'
ret = strData.removeprefix('<EOD>')
print(ret)
```
