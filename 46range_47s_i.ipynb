```python
#46순차적 정수열 만들기: range()
```

```python
intList = range(10)
print(list(intList))
```

```python
intList = range(10, 20)
print(list(intList))
```

```python
intList = range(1, 20, 3)
print(list(intList))
```

```python
intList = [x for x in range(1,10,2)]
print(list(intList))
```
