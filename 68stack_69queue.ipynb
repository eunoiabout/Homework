```python
#68 리스트로 스택 구현하기(stack)
```

```python
mystack = []
```

```python
def putData(data):
    mystack.append(data)
```

```python
def popData():
    if len(mystack) == 0:
        return None
    return mystack.pop()
```

```python
putData('사과')
putData([1,2,3])
putData(3.141592)
print(f'스택상태:{mystack}')
```

```python
ret = popData()
while ret != None:
   print(f'스택에서 데이터 추출{ret}')
   print(f'스택상태:{mystack}')
   print('-----------------')
   ret = popData()
```

```python
#69 리스트로 큐 구현하기 queue
```

```python
import collections
myqueue = collections.deque([])
```

```python
def putData(data):
    myqueue.append(data)
```

```python
def popData():
  if len(myqueue) == 0:
    return None
  return myqueue.popleft()
```

```python
putData('사과')
putData([1,2,3])
putData(3.141592)
print(f'큐상태:{myqueue}')
```

```python
ret = popData()
while ret != None:
  print(f'큐에서 데이터 추출:{ret}')
  print(f'큐상태:{myqueue}')
  print('-----------------')
  ret = popData()
```
