```python
#73 문자 코드값 구하기:ord()
```

```python
while True:
   ch = input('문자1개를 입력하세요[끝내려면 엔터를 치세요]: ')
   if len(ch) ==0:
      break
   ch = ch[0]
   code =ord(ch)
   print(f'-->문자 "{ch}"의  코드값:{hex(code)}')
```
