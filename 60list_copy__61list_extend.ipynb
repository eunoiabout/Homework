```python
#60 리스트  복사하기: list.copy()
```

```python
fruits = ['사과','배','오렌지','수박','참외']
buffer = fruits.copy()
del buffer[-1]
print(f'원본 리스트: {fruits}')
print(f'복사본 리스트: {buffer}')
```

```python
#61 리스트 확정하기: list.extend()
```

```python
world = []
asia = ['한국','중국','일본']
europe = ['독일','프랑스','영국']
america = ['미국','캐나다','멕시코']
```

```python
world.extend(asia)
print(world)
```

```python
world.extend(europe)
print(world)
world.extend(america)
print(world)
```

```python
world.clear()
world += asia
world += europe
print(world)
```
