```python
#16for문: for~continue~break
```

```python
strData = 'abcdef'
for c in strData:
  print(c, end='-')
```

```python
listData = [1,2,3,4,5]
for idx in listData:
  print(idx,end='-')
```

```python
ascii_codes = {'a':97,'b':98,'c':99}
for code in ascii_codes.items():
  print(code)
```

```python
for i in range(10):
  print(i,end='#')
```

```python
for i in [1,2,3,4,5]:
  print(i)
  if i < 2:
    continue
  else:
    break
```

```python
for i in [1,2,3,4,5]:
  print(i)
  if i == 2:
    break
```

```python
#17for문: for~else
```

```python
for x in [1,2,3]:
  print(x)
else:
  print('perfect!')
```

```python
for x in [1,2,3]:
  print(x)
  if x > 1:
    break
else:
  print('perfect!')
```
