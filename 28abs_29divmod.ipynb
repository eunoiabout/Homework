```python
#29몫과 나머지 구하기: dovmod()
```

```python
x = 11113
y = 23
ret1 =  x // y
ret2 = x % y
print(f'{x}을 {y}로 나누면 몫이 {ret1}, 나머지가 {ret2}입니다.')
```

```python
x = 11113
y = 23
ret1, ret2 = divmod(x, y)
print(f'{x}을 {y}로 나누면 몫이 {ret1}, 나머지가 {ret2}입니다.')
```
