```python
#20match문:match~case
```

```python
http_status = 400
match http_status:
  case 400:
    print('Bad Request')
  case 401:
    print('Unauthorized')
  case 403:
    print('Forbidden')
  case 404:
    print('Not Found')
  case _:
    print('기타 인터넷 문제로 뭔가 잘못됨.')
```

```python
http_status = 302
match http_status:
  case 200|202|203:
    print('Sucess')
  case 301|302|303:
    print('Redirection')
  case 400|401|403|404:
    print('ClientErrors')
  case _:
    print('기타 HTTP응답 코드')

```
