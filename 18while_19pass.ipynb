```python
x = 0
while x < 10:
  x = x + 1
  if x < 3:
    continue

  print(x)
  if x > 4:
    break
```

```python
n = 1
total = 0
while True:
  total = total + n
  if total > 100000:
    print('1+2+...+n 값이 10만 보다 커지게 되는 n:',n)
    print('이 때 총합은',total)
    break
  n = n + 1
```
