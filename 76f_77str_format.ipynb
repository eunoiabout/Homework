```python
#76 문자열 포맷팅 방법:f''
```

```python
title = '원주율'
pi = 3.141592
print(f'{title}의 근사값은 {pi}입니다.')
```

```python
print(f'{title}의 근사값은 {pi:.2f}입니다.')
```

```python
city = '서울'
population = 8789911
print(f'{city}:{population:,d} 명')
```

```python
print(f'{city}:{population=:,d} 명')
```

```python
#77 문자열 포맷팅 방법:str.format()
```

```python
pi = 3.141592
text = '{}의 근사값은 {}입니다.'.format('원주율', pi)
print(text)
```

```python
print('{}의 근사값은 {:.2f}입니다.'.format('원주율', pi))
```

```python
population = 8789911
text = '{}:{:,d}명'.format('서울', population)
print(text)
```
