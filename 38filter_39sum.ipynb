```python
#38특정 조건을 만족하는 요소만 추출하기: filter()
```

```python
def getPrime(x):
    if x == 2:
      return True
    if x <= 1 or x % 2 == 0:
      return False

    for i in range(3, int(x**(1/2)) + 1,2):
        if x % i == 0:
          return False
    else:
        return True
```

```python
intList = [x for x in range(1, 101)]
ret = filter(getPrime, intList)
for p in ret:
  print(p, end=' ')
```

```python
numList = [1,3,5,7,9]
ret = sum(numList)
print(ret)
```

```python
ret = sum(numList, 10)
print(ret)
```

```python
randomList = [1,2, 'SAMSUNG', 'LG']
# Filter out non-integer elements before summing
numeric_list = [item for item in randomList if isinstance(item, (int, float))]
ret = sum(numeric_list)
print(ret)
```
