```python
#64리스트에서 특정 위치 요소 추출 후 삭제: list.pop()
```

```python
numbers = [14,5,18,45,10]
for _ in range(len(numbers)):
    ret = numbers.pop()
    print(f'pop()으로 추출된 값: {ret}, numbers: {numbers}')
```

```python
numbers = [14,5,18,45,10]
ret = numbers.pop(3)
print(f'pop(0)으로 추출된 값: {ret}, numbers: {numbers}')
```
