```python
#80 문자열이 숫자로만 구성되었는지 확인: str.isdecimal(),str.isdight(),str.isnumeric()
```

```python
numbers = ['1004', '-1', '3.14','IX','⅞','5^2','50%']
```

```python
for number in numbers:
    print(f'"{number}"\t-->{number.isdecimal()}')
```

```python
for number in numbers:
   print(f'"{number}"\t-->{number.isdigit()}')
```

```python
for number in numbers:
    print(f'"{number}"\t-->{number.isnumeric()}')
```
