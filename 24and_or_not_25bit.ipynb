```python
#24논리연산자:and, or, not
```

```python
x = 1; y = 2
strData1 = 'Hello';strData2 = 'python'
flag1 = x == y
flag2 = x < y
flag3 = strData1 != strData2
flag4 = strData1 > strData2

print(flag1 and flag2)
print(flag2 and flag3)
print(flag3 and flag4)
print(flag1 and flag4)

print(not flag1)
print(not flag2)
```

```python
#25비트연산자: &,|,^.>>.<<
```

```python
bit1 = 0x61; bit2 = 0x62
print(hex(bit1 & bit2))
print(hex(bit1 | bit2))
print(hex(bit1 ^ bit2))
print(hex(bit1 >> bit2))
print(hex(bit1 << bit2))
```

```python
byte1 = 0x6b

lower_4bit = byte1 & 0x0f

upper_4bit = (byte1 >> 4) & 0x0f

print(hex(lower_4bit))
print(hex(upper_4bit))
```
