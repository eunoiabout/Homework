```python
#40반복 가능 자료를 리스트로 변환: list()
```

```python
tupleData = (1,2,3,4,5)
print(list(tupleData))
```

```python
strData = 'I love you!'
print(list(strData))
```

```python
def MyRange(n):
  current = 0
  while current < n:
    yield current
    current += 1

intList = MyRange(5)
print(list(intList))
```

```python
#41모든 요소들에 대해 함수 리턴값 얻기: map()
```

```python
def myFunc(x):
    return x ** 2

results = map(myFunc, [0,1,2,3])
print(list(results))
```

```python
results = map(lambda x:x ** 2, range(4))
print(list(results))
```

```python
x = [1,2,3,4,5]
y = [10, 9, 8, 7, 6]
ret = map(lambda x, y:x ** 2 + y, x, y)
print(list(ret))
```
