```python
#82 문자열을 구분자로 분리하기: str.split()
```

```python
logData = '홍길동;남자;18세;도적;초능력자'
profile = logData.split(';')
print(profile)
```

```python
text = '나는 파이썬을 사랑합니다.'
print(text.split())
```

```python
profile = '이름:홍길동 나이:18세 성별:남자 국적:조선 직업:도적 특기:초능력'
for data in profile.split():
  key, val = data.split(':')
  print(f'{key} --> {val}')
```

```python
#83 문자열에서 좌우 문자/공백 제거하기:str.strip(),str.lstrip(),str.rstrip()
```

```python
text = ' 이 문장 양쪽에는 공백이 있습니다. '
print(f'<{text.strip()}>')
print(f'<{text.lstrip()}>')
print(f'<{text.rstrip()}>')
```

```python
url = 'www.python.com'
print(url.strip('cmow.'))
```

```python
print(url.lstrip('comw.'))
print(url.rstrip('comw.'))
```
