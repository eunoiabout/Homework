```python
#3210진수를 6진수로 변환: hex()
```

```python
h1 = hex(97)
h2 = hex(98)
print(h1)
print(h2)
```

```python
print(h1 + h2)
```

```python
hexstr1 = '0x11'
hexstr2 = '0x2a'
int1 = int(hexstr1, base = 16)
int2 = int(hexstr2, base = 16)

hexAnswer = hex(int1 + int2)
print(f'{hexstr1} + {hexstr2} = {hexAnswer}')
```

```python
#3310진수를 2진수로 변환: bin()
```

```python
b1 = bin(97)
b2 = bin(98)
print(b1)
print(b2)
```

```python
print(b1 +b2)
```

```python
binstr1 = '0b1011'
binstr2 = '0b1001'
int1 = int(binstr1, base = 2)
int2 = int(binstr2, base = 2)

binAnswer = bin(int1 + int2)
print(f'{binstr1} + {binstr2} = {binAnswer}')
```
