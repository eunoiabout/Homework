```python
#74 코드값에 대응하는 문자 얻기:chr()
```

```python
while True:
  code = input('0x로 시작하는 16진수 문자 코드값을 입력하세요[끝내려면 엔터를 치세요];')
  if len(code) ==0:
     break
  unicode = int(code,16)
  try:
     ch = chr(unicode)
     print(f'-->코드값:{code}, 문자:"{ch}"')
  except ValueError:
     print(f'코드값{code}에 해당하는 문자가 존재하지 않습니다.')
```
